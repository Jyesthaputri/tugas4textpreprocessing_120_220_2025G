```python
import pandas as pd

df = pd.read_csv('dataset_instagram-comment-scraper.csv')

for i, row in enumerate(df['text'], 1):
    text_clean = str(row).replace('\n', ' ')
    print(f'[{i:03d}] {text_clean}')
    print('-' * 80)
```

```python
import re
def remove_mention_hastag(text):
    text = re.sub(r'#\w+', '', text)
    text = re.sub(r'@[A-Za-z0-9_]+', '', text)
    return text

df['1_hastag_mention_removal'] = df['text'].apply(remove_mention_hastag)

print("1. Hashtag & mention removal")
display(df['1_hastag_mention_removal'].head(10))
```

```python
def lower_casing(text):
    return str(text).lower()

df['2_lower'] = df['1_hastag_mention_removal'].apply(lower_casing)

print("2. Lower casing")
display(df['2_lower'].head(10))
```

```python
import emoji

def demojize_text(text):
    return emoji.demojize(str(text), delimiters=(' emoji_', ' '), language='id')

df['3_autoemoji'] = df['2_lower'].apply(demojize_text)

print("3. Removal of emoji")
display(df['3_autoemoji'].head(10))
```

```python
import re

def remove_punctuations(text):
    text = re.sub(r'[^\w\s]', ' ', str(text))
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['4_punct'] = df['3_autoemoji'].apply(remove_punctuations)

print("4. Removal of Punctuations")
display(df['4_punct'].head(10))
```

```python
import re

def normalize_text(text):
    text = re.sub(r'(\w)\1{2,}', r'\1', text)
    return text

df['5_normalization'] = df['4_punct'].apply(normalize_text)

print("5. Text Normalization")
display(df['5_normalization'].head(10))
```

```python
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

factory = StopWordRemoverFactory()
stop_words = set(factory.get_stop_words())

kata_negasi = {'tidak', 'bukan', 'belum', 'jangan'}
stop_words = stop_words - kata_negasi
stop_words.update({'berada', 'mem','begini','gini', 'sedang', 'lah', 'tuh', 'yakali'
      'hiya', 'tara', 'makjrengg', 'pas', 'ku', 'nya','mu', 'si', 'kok', 'euy', 'nih', 'heh'
      'yakali', 'wahh', "nii", 'keadaan', 'mau', 'pak', 'jadi', 'seperti', 'sekali', 'cuma', 'memang',
      'buat', 'bahkan', 'terlalu', 'mudah', 'banget', 'selalu', 'sama', 'selalu', 'semua',
      'diantaranya', 'melalui', 'sana', 'sekitar', 'sebentar', 'bakal', 'kalau', 'bareng'
      })

def remove_stopwords(text):
    if not isinstance(text, str):
        return ""
    words = str(text).split()
    filtered = [word for word in words if word.lower() not in stop_words]
    return ' '.join(filtered)

df['8_stopword'] = df['7_chat_words'].apply(remove_stopwords)
df['8_stopword'] = df['8_stopword'].apply(
    lambda x: re.sub(r'\s+', ' ', x).strip()
)

print('8. Removal of Stopwords')
display(df['8_stopword'].head(10))
```

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

factory = StemmerFactory()
dictionary = factory.get_words()

custom_words = {
    'rasa', 'pemerintah', 'selimut', 'menteri', 'ingin',
    'tangis', 'puas', 'pemadam', 'anak', 'pelaku'
    }

for word in custom_words:
    dictionary.append(word)

stemmer = factory.create_stemmer()
STEM_OVERRIDE_MAP = {
    'terasa': 'rasa', 'berasa': 'rasa', 'merasa': 'rasa',
    'pemerintah': 'pemerintah',
    'selimut': 'selimut', 'diselimuti': 'selimut',
    'menteri': 'menteri', 'kementrian': 'kementrian',
    'pengen': 'ingin', 'pengin': 'ingin', 'menginginkan': 'ingin',
    'nangis': 'tangis', 'menangis': 'tangis', 'tangisan': 'tangis',
    'puas': 'puas', 'memuaskan': 'puas', 'kepuasan': 'puas',
    'pemadam': 'pemadam',
    'anak': 'anak', 'anakan': 'anak',
    'pelaku': 'pelaku',
    'paruparu': 'paruparu',
    'tibatiba': 'tibatiba',
    'wapresnya':'wapres'
    }

def stem_per_word(text):
    words = str(text).split()
    stemmed_words = []

    for word in words:
        if word.startswith('emoji_'):
            stemmed_words.append(word)
        elif word in STEM_OVERRIDE_MAP:
            stemmed_words.append(STEM_OVERRIDE_MAP[word])
        else:
            stemmed_words.append(stemmer.stem(word))

    return ' '.join(stemmed_words)

df['9_stemming'] = df['8_stopword'].apply(stem_per_word)
df.to_csv('dataset_komentar_clean.csv', index=False, encoding='utf-8-sig')

print("9. Stemming")
display(df['9_stemming'].head(10))
```
